# How to Use `git clone` in WSL (Windows Subsystem for Linux)

To use `git clone` in WSL, open your Linux terminal and follow these
steps to set up and run your command.

## Step 1: Open WSL

Launch your Linux terminal (e.g., Ubuntu on WSL) from your Windows Start
menu, or open Windows Command Prompt / PowerShell and type:

``` bash
wsl
```

## Step 2: Install Git (If Not Already Installed)

WSL distributions sometimes come with Git pre-installed. You can check
if it's available by running:

``` bash
git --version
```

If it returns a `command not found` error, install Git using your
distribution's package manager:

-   **Ubuntu / Debian:**

    ``` bash
    sudo apt update
    sudo apt install git
    ```

-   **Fedora:**

    ``` bash
    sudo dnf install git
    ```

## Step 3: Run `git clone`

Navigate to the directory where you want to download your repository
using the `cd` command (for instance, your Linux home directory), and
then run the `git clone` command with your repository URL:

``` bash
cd ~
git clone https://github.com/username/repository-name.git
```

Replace `username/repository-name` with the actual GitHub username and
repository name.

## Useful WSL Tips

-   **Performance:** Clones placed inside your WSL home directory (`~`)
    live within the Linux file system, giving you much faster
    performance for development tools and packages.
-   **Access from Windows:** You can browse and edit your WSL files
    directly from Windows File Explorer by typing `\\wsl$` in the
    address bar.
-   **Authentication:** If you are cloning private repositories, set up
    your Personal Access Token (PAT) or configure SSH keys inside your
    WSL terminal just like on standard Linux.


# Change directory to the github folder clone in your system for futher operations

``` bash
cd Repository-Name
code .
```

## Important: Virtual Environment Was Committed

The commit output shows that the `project/` virtual environment was
included in Git. Virtual environments should generally not be committed;
they can be recreated from `requirements.txt`.

To stop tracking it while keeping your local environment:

1.  Add the following to `.gitignore`:

    ``` gitignore
    # Python virtual environments
    project/
    .venv/
    venv/

    # Python cache
    __pycache__/
    *.py[cod]

    # Environment variables and secrets
    .env
    ```

2.  Remove the virtual environment from Git tracking:

    ``` bash
    git rm -r --cached project/
    ```

3.  Commit and push the cleanup:

    ``` bash
    git add .gitignore
    git commit -m "Stop tracking virtual environment"
    git push origin main
    ```

This cleanup removes `project/` from the latest version of the
repository without deleting your local virtual environment. The earlier
commit remains in Git history.
